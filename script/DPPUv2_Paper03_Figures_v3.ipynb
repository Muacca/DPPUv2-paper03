# Paper03: spin-0, spin-1 and spin-2 figures

Render ten figures from saved datasets, or set RECOMPUTE=True for explicit fixed-grid recomputation. Each run writes to new time-stamped folders next to the script directory (notebook-figures-YYYYMMDD-HHMMSS, and notebook-data-... when recomputing). The figures distinguish static minima, restricted circle actions, derivative cross terms and the coupled periodic sequence.


```python
from pathlib import Path
import sys
HERE = Path.cwd().resolve()
SCRIPT = HERE if (HERE/'paper_results').is_dir() else HERE/'script'
if not (SCRIPT/'paper_results').is_dir():
    raise RuntimeError('Run from the repository root or script directory.')
sys.path.insert(0,str(SCRIPT))
RECOMPUTE = False
DATA = SCRIPT.parent/'data/revised'
```

```python
from datetime import datetime
STAMP = datetime.now().strftime('%Y%m%d-%H%M%S')
OUTPUT = SCRIPT.parent/f'notebook-figures-{STAMP}'
NEW_DATA = SCRIPT.parent/f'notebook-data-{STAMP}'
if RECOMPUTE:
    from paper_results.data import recompute
    DATA = recompute(NEW_DATA)
from paper_results.figures import render
paths = render(DATA,OUTPUT)
print('Figures written to', OUTPUT)
from IPython.display import display,Image
for path in paths:
    if path.suffix == '.png': display(Image(filename=str(path)))
```
